19. Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o
ângulo entre eles (produto escalar normalizado).

```python
import numpy as np

arr1 = np.random.randint(1, 51, size=20)
arr2 = np.random.randint(1, 51, size=20)
print("Array 1:", arr1)
print("Array 2:", arr2)

produto_escalar = np.dot(arr1, arr2)
print("Produto escalar:", produto_escalar)

norma1 = np.linalg.norm(arr1)
norma2 = np.linalg.norm(arr2)

cos_theta = produto_escalar / (norma1 * norma2)

angulo_rad = np.arccos(np.clip(cos_theta, -1, 1))
angulo_graus = np.degrees(angulo_rad)
print("Ângulo entre os vetores:", round(angulo_graus, 2), "graus")
```

1. Calcule a integral da função f(x) = x² sin(x) no intervalo de 0 a π.

```python
import numpy as np
from scipy.integrate import quad

def f(x):
    return x**2 * np.sin(x)

resultado, erro = quad(f, 0, np.pi)

print(f"Integral de x² * sin(x) de 0 a π: {resultado:.6f}")
print(f"Erro estimado: {erro:.2e}")
```

2. Resolva a equação diferencial (dy/dx) = x − y, com y(0) = 1, no intervalo de 0
a 10.

```python
import numpy as np
from scipy.integrate import solve_ivp

def dy_dx(x, y):
    return x - y[0]

solucao = solve_ivp(dy_dx, [0, 10], [1], t_eval=np.linspace(0, 10, 100))

print("Solução da EDO dy/dx = x - y, y(0) = 1:")
print(f"{'x':>8}  {'y':>10}")
for x, y in zip(solucao.t[::10], solucao.y[0][::10]):
    print(f"{x:8.4f}  {y:10.6f}")
```

3. Encontre as raízes da função f(x) = x⁴ − 3x³ + 2 utilizando SciPy.

```python
import numpy as np
from scipy.optimize import fsolve

def f(x):
    return x**4 - 3*x**3 + 2

chutes_iniciais = [-1, 0.5, 1.5, 3]
raizes_encontradas = set()

print("Raízes da função f(x) = x⁴ - 3x³ + 2:")
for chute in chutes_iniciais:
    raiz = fsolve(f, chute, full_output=True)
    x_raiz = raiz[0][0]
    info = raiz[1]

    if abs(f(x_raiz)) < 1e-10:
        raiz_arredondada = round(x_raiz, 8)
        if raiz_arredondada not in raizes_encontradas:
            raizes_encontradas.add(raiz_arredondada)
            print(f"  x = {x_raiz:.8f}  |  f(x) = {f(x_raiz):.2e}")

print(f"\nTotal de raízes encontradas: {len(raizes_encontradas)}")
```

4. Use a função de otimização de SciPy para maximizar f(x) = −x²+4x+1.

```python
import numpy as np
from scipy.optimize import minimize

def f_negativa(x):
    return -(-x**2 + 4*x + 1)

resultado = minimize(f_negativa, x0=0)

x_max = resultado.x[0]
f_max = -resultado.fun

print("Maximização de f(x) = -x² + 4x + 1:")
print(f"  x ótimo: {x_max:.6f}")
print(f"  f(x) máximo: {f_max:.6f}")
print(f"  Convergiu: {resultado.success}")
```

5. Calcule a transformada de Fourier de um sinal composto de sin(2πt) e
cos(4πt).

```python
import numpy as np
from scipy.fft import fft

N = 1000
T = 1.0 / N
t = np.linspace(0, 1, N, endpoint=False)

sinal = np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t)

transformada = fft(sinal)
frequencias = np.fft.fftfreq(N, T)

print("Transformada de Fourier do sinal sin(2πt) + cos(4πt):")
print(f"  Número de amostras: {N}")
print(f"\n  Frequências dominantes (|amplitude| > 1):")
for i in range(N // 2):
    amplitude = 2.0 / N * np.abs(transformada[i])
    if amplitude > 0.5:
        print(f"    Frequência: {frequencias[i]:.1f} Hz  |  Amplitude: {amplitude:.4f}")
```

9. Ajuste uma curva quadrática aos pontos (1, 2), (2, 4), (3, 5), (4, 7), (5, 8).

```python
import numpy as np
from scipy.optimize import curve_fit

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 7, 8])

def modelo_quadratico(x, a, b, c):
    return a * x**2 + b * x + c

parametros, covariancia = curve_fit(modelo_quadratico, x, y)

a, b, c = parametros

print("Ajuste de curva quadrática f(x) = ax² + bx + c:")
print(f"  a = {a:.6f}")
print(f"  b = {b:.6f}")
print(f"  c = {c:.6f}")
print(f"\n  Equação ajustada: f(x) = {a:.4f}x² + {b:.4f}x + {c:.4f}")

print(f"\nComparação (dados originais vs ajustados):")
print(f"  {'x':>4}  {'y real':>8}  {'y ajustado':>11}")
for xi, yi in zip(x, y):
    y_pred = modelo_quadratico(xi, a, b, c)
    print(f"  {xi:4.0f}  {yi:8.1f}  {y_pred:11.4f}")
```

16. Resolva a equação diferencial de segunda ordem (d²y/
dx²) +y = 0, com condições
iniciais y(0) = 1 e y′(0) = 0.

```python
import numpy as np
from scipy.integrate import solve_ivp

def sistema(x, Y):
    y1, y2 = Y
    dy1_dx = y2
    dy2_dx = -y1
    return [dy1_dx, dy2_dx]

y0 = [1, 0]

t_span = [0, 4 * np.pi]
t_eval = np.linspace(0, 4 * np.pi, 200)

solucao = solve_ivp(sistema, t_span, y0, t_eval=t_eval, method='RK45')

print("EDO de 2ª ordem: d²y/dx² + y = 0")
print("Condições iniciais: y(0) = 1, y'(0) = 0")
print("Solução analítica: y(x) = cos(x)")
print(f"\n{'x':>8}  {'y numérico':>12}  {'cos(x)':>10}  {'erro':>10}")
for i in range(0, len(solucao.t), 20):
    x = solucao.t[i]
    y_num = solucao.y[0][i]
    y_analitico = np.cos(x)
    erro = abs(y_num - y_analitico)
    print(f"{x:8.4f}  {y_num:12.8f}  {y_analitico:10.8f}  {erro:10.2e}")
```

19. Realize a transformada de Fourier de um sinal discreto composto de senos
e cossenos.

```python
import numpy as np
from scipy.fft import fft

N = 500
fs = 500
t = np.arange(N) / fs

f1, f2, f3 = 10, 25, 50
sinal = 3 * np.sin(2 * np.pi * f1 * t) + 1.5 * np.cos(2 * np.pi * f2 * t) + 2 * np.sin(2 * np.pi * f3 * t)

transformada = fft(sinal)
frequencias = np.fft.fftfreq(N, 1.0 / fs)
amplitudes = 2.0 / N * np.abs(transformada[:N // 2])
freq_positivas = frequencias[:N // 2]

print("Transformada de Fourier do sinal discreto:")
print(f"  Sinal: 3·sin(2π·{f1}t) + 1.5·cos(2π·{f2}t) + 2·sin(2π·{f3}t)")
print(f"  Amostras: {N}  |  Frequência de amostragem: {fs} Hz")

print(f"\n  Frequências dominantes (amplitude > 0.5):")
for f, amp in zip(freq_positivas, amplitudes):
    if amp > 0.5:
        print(f"    Frequência: {f:.1f} Hz  |  Amplitude: {amp:.4f}")
```

20. Resolva um sistema de equações lineares com três variáveis utilizando
SciPy.

```python
import numpy as np
from scipy.linalg import solve

A = np.array([[2,  3, -1],
              [4,  1,  2],
              [-1, 2,  3]])

b = np.array([1, 2, 3])

solucao = solve(A, b)

print("Sistema de equações lineares com 3 variáveis:")
print("  2x + 3y -  z = 1")
print("  4x +  y + 2z = 2")
print("  -x + 2y + 3z = 3")
print(f"\nSolução:")
print(f"  x = {solucao[0]:.6f}")
print(f"  y = {solucao[1]:.6f}")
print(f"  z = {solucao[2]:.6f}")

# Verificação
print(f"\nVerificação:")
resultado = A @ solucao
print(f"  2x + 3y - z  = {resultado[0]:.6f}  (esperado: 1)")
print(f"  4x + y  + 2z = {resultado[1]:.6f}  (esperado: 2)")
print(f"  -x + 2y + 3z = {resultado[2]:.6f}  (esperado: 3)")
```

11. Preencha os valores nulos de uma coluna numérica com a mediana dessa
coluna.

```python
import pandas as pd
import numpy as np

dados = {
    'Produto': ['Arroz', 'Feijão', 'Macarrão', 'Açúcar', 'Café'],
    'Preço': [22.90, np.nan, 5.49, np.nan, 18.75],
    'Quantidade': [50, 80, 120, 90, 60]
}

df = pd.DataFrame(dados)

print("DataFrame original (com valores nulos):")
print(df)

mediana_preco = df['Preço'].median()
print(f"\nMediana da coluna Preço: {mediana_preco}")

df['Preço'] = df['Preço'].fillna(mediana_preco)

print("\nDataFrame com valores nulos preenchidos pela mediana:")
print(df)

```

12. Verifique quais colunas de um DataFrame contêm valores duplicados.

```python
import pandas as pd

dados = {
    'Produto': ['Arroz', 'Feijão', 'Arroz', 'Açúcar', 'Café'],
    'Preço': [22.90, 8.50, 22.90, 4.99, 18.75],
    'Quantidade': [50, 80, 50, 90, 60],
    'Categoria': ['Grãos', 'Grãos', 'Grãos', 'Doces', 'Bebidas']
}

df = pd.DataFrame(dados)

print("DataFrame:")
print(df)

print("\nVerificação de duplicatas por coluna:")
for coluna in df.columns:
    tem_duplicados = df[coluna].duplicated().any()
    if tem_duplicados:
        print(f"  '{coluna}' → contém valores duplicados")
    else:
        print(f"  '{coluna}' → sem valores duplicados")

```

14. Mescle dois DataFrames baseados em uma coluna comum chamada ID

```python
import pandas as pd

dados_clientes = {
    'ID': [1, 2, 3, 4, 5],
    'Nome': ['Ana', 'Bruno', 'Carlos', 'Diana', 'Eduardo'],
    'Cidade': ['São Paulo', 'Rio de Janeiro', 'Curitiba', 'Belo Horizonte', 'Salvador']
}

dados_pedidos = {
    'ID': [1, 2, 3, 4, 5],
    'Produto': ['Notebook', 'Mouse', 'Teclado', 'Monitor', 'Webcam'],
    'Valor': [3500.00, 89.90, 150.00, 1200.00, 95.00]
}

df_clientes = pd.DataFrame(dados_clientes)
df_pedidos = pd.DataFrame(dados_pedidos)

print("DataFrame Clientes:")
print(df_clientes)

print("\nDataFrame Pedidos:")
print(df_pedidos)

df_mesclado = pd.merge(df_clientes, df_pedidos, on='ID')

print("\nDataFrame Mesclado:")
print(df_mesclado)
```

8. Plote um gráfico de linha com intervalos de confiança (sombramento) para
a função y = sin(x).

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 200)
y = np.sin(x)
erro = 0.2

plt.plot(x, y, color='blue', linewidth=2, label='y = sin(x)')
plt.fill_between(x, y - erro, y + erro, color='blue', alpha=0.2, label='Intervalo de confiança')
plt.title('Gráfico de sin(x) com Intervalo de Confiança')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
```

18. Crie um gráfico de barras com valores positivos e negativos e use diferentes
cores para cada tipo de valor.

```python
import numpy as np
import matplotlib.pyplot as plt

categorias = ['A', 'B', 'C', 'D', 'E', 'F']
valores = [15, -10, 25, -5, 30, -20]
cores = ['#2ecc71' if v >= 0 else '#e74c3c' for v in valores]

plt.bar(categorias, valores, color=cores, edgecolor='black')
plt.title('Barras com Valores Positivos e Negativos')
plt.xlabel('Categorias')
plt.ylabel('Valores')
plt.axhline(y=0, color='black', linewidth=0.8)
plt.grid(True, axis='y')
plt.show()
```

19. Plote um gráfico de linha com duas séries e adicione uma legenda para
diferenciá-las.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 200)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1, color='blue', linestyle='-', linewidth=2, label='sin(x)')
plt.plot(x, y2, color='red', linestyle='--', linewidth=2, label='cos(x)')
plt.title('Duas Séries com Legenda')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
```

20. Adicione uma grade com espaçaamento personalizado a um gráfico de linha
para y = x².

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 200)
y = x ** 2

fig, ax = plt.subplots()
ax.plot(x, y, color='darkorange', linewidth=2)
ax.set_title('Gráfico de y = x² com Grade Personalizada')
ax.set_xlabel('x')
ax.set_ylabel('y')

ax.set_xticks(np.arange(-10, 11, 2))
ax.set_yticks(np.arange(0, 101, 10))
ax.grid(True, which='major', linestyle='-', linewidth=0.8, color='gray', alpha=0.7)

ax.set_xticks(np.arange(-10, 11, 1), minor=True)
ax.set_yticks(np.arange(0, 101, 5), minor=True)
ax.grid(True, which='minor', linestyle=':', linewidth=0.4, color='lightgray', alpha=0.5)

plt.show()
```
